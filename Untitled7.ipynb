```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
from scipy import stats
from sklearn.metrics import mean_absolute_error, mean_squared_error
from statsmodels.tsa.stattools import adfuller
from statsmodels.graphics.tsaplots import plot_acf, plot_pacf
```

```python
data = pd.read_csv('skab_valve_1.csv', sep=';', encoding='latin-1')
data['datetime'] = pd.to_datetime(data['datetime'], dayfirst=True)
data = data.sort_values('datetime')
```

```python
thermocouple = data['Thermocouple']
```

```python
plt.figure(figsize=(12, 5))
plt.plot(data['datetime'], thermocouple, linewidth=2)
plt.title('Данные термопары')
plt.xlabel('Время')
plt.ylabel('Температура (°C)')
plt.grid(True)
plt.show()
```

анализ временного ряда

```python
print(f"{len(thermocouple)} точек")
print(f"Среднее: {thermocouple.mean():.2f}°C")
```

```python
x = np.arange(len(thermocouple))
slope, _, _, p_value, _ = stats.linregress(x, thermocouple.values)
```

```python
print(f"тренд: {'есть' if p_value < 0.05 else 'нет'}")
print("сезонность слабая, ряд стационарен")
```

```python
plt.figure(figsize=(12, 4))
plt.plot(data['datetime'].head(500), thermocouple.head(500), linewidth=2)
plt.title('Фрагмент данных (первые 500 точек)')
plt.xlabel('Время')
plt.ylabel('Температура (°C)')
plt.grid(True)
plt.show()
```

```python
#разделение на train/test (50/50)
train_size = int(len(thermocouple) * 0.5)
train_data = thermocouple.iloc[:train_size]
test_data = thermocouple.iloc[train_size:]

print(f"обучающая выборка: {len(train_data)}")
print(f"тестовая выборка: {len(test_data)}")
```

```python
plt.figure(figsize=(12, 5))
plt.plot(data['datetime'].iloc[:train_size], train_data, linewidth=2, label='Train', color='blue')
plt.plot(data['datetime'].iloc[train_size:], test_data, linewidth=2, label='Test', color='red')
plt.title('Разделение данных на Train/Test')
plt.xlabel('Время')
plt.ylabel('Температура (°C)')
plt.legend()
plt.grid(True)
plt.show()
```

```python
print("статистика по выборкам:")
print(f"train - {train_data.mean():.2f}°C, Std: {train_data.std():.2f}")
print(f"test - {test_data.mean():.2f}°C, Std: {test_data.std():.2f}")
```

для данных термопары промышленного оборудования лучше выьрать метрики MAE (Mean Absolute Error) и RMSE (Root Mean Square Error). MAE показывает среднюю абсолютную ошибку в единицах измерения (°C), что делает её легко интерпретируемой. RMSE учитывает большие ошибки и штрафует за выбросы, что важно для обнаружения аномальных отклонений

```python
#бейзлайн наивный
last_train_value = train_data.iloc[-1]
naive_predictions = np.full(len(test_data), last_train_value)

print(f"последнее значение из train: {last_train_value:.4f}°C")
print(f"прогноз для test: постоянное значение {last_train_value:.4f}°C")

plt.figure(figsize=(12, 5))
plt.plot(data['datetime'].iloc[train_size:], test_data, linewidth=2, label='Фактические значения', color='blue')
plt.plot(data['datetime'].iloc[train_size:], naive_predictions, linewidth=2, label='Наивный прогноз', color='red', linestyle='--')
plt.title('Бейзлайн: наивный прогноз vs Фактические значения')
plt.xlabel('Время')
plt.ylabel('Температура (°C)')
plt.legend()
plt.grid(True)
plt.show()
```

```python
print(f"test_data has NaN: {test_data.isna().any()}")
print(f"naive_predictions has NaN: {np.isnan(naive_predictions).any()}")

test_data_clean = test_data.fillna(method='ffill')
naive_predictions_clean = np.nan_to_num(naive_predictions, nan=test_data_clean.mean())

print("ОЦЕНКА КАЧЕСТВА НАИВНОГО ПРОГНОЗА:")
mae_naive = mean_absolute_error(test_data_clean, naive_predictions_clean)
rmse_naive = np.sqrt(mean_squared_error(test_data_clean, naive_predictions_clean))

print(f"MAE: {mae_naive:.4f}°C")
print(f"RMSE: {rmse_naive:.4f}°C")
```

```python
#тест Дики-Фуллера
result = adfuller(thermocouple.dropna())

print(f"ADF статистика: {result[0]:.4f}")
print(f"p-value: {result[1]:.4f}")

if result[1] < 0.05:
    print("ряд стационарен")
else:
    print("ряд нестационарен")
```

```python
#ACF и PACF
from statsmodels.graphics.tsaplots import plot_acf, plot_pacf

plt.figure(figsize=(12, 6))

plt.subplot(1, 2, 1)
plot_acf(thermocouple.dropna(), ax=plt.gca(), lags=40)
plt.title('ACF')

plt.subplot(1, 2, 2)
plot_pacf(thermocouple.dropna(), ax=plt.gca(), lags=40)
plt.title('PACF')

plt.tight_layout()
plt.show()

print("ACF")
print("медленное затухание корреляции подтверждает нестационарность ряда")
print("PACF")
print("резкий обрыв после нескольких первых лагов")
print("значит")
print("ряд нестационарный")
print("явная сезонность отсутствует")
```

```python
#дифференцирование ряда для придания стационарности
thermocouple_diff = thermocouple.diff().dropna()

plt.figure(figsize=(12, 8))

plt.subplot(2, 1, 1)
plt.plot(thermocouple)
plt.title('исходный ряд (нестационарный)')
plt.grid(True)

plt.subplot(2, 1, 2)
plt.plot(thermocouple_diff)
plt.title('продифференцированный ряд (1 порядок)')
plt.grid(True)

plt.tight_layout()
plt.show()

result_diff = adfuller(thermocouple_diff)

print("проверка после диф:")
print(f"ADF статистика: {result_diff[0]:.4f}")
print(f"p-value: {result_diff[1]:.4f}")

if result_diff[1] < 0.05:
    print("ряд стал стационарным")
else:
    print("ряд все еще нестационарен")

# Статистики для сравнения
print(f"сравнение")
print(f"исходный ряд: среднее = {thermocouple.mean():.4f}, std = {thermocouple.std():.4f}")
print(f"дифференцированный ряд: среднее = {thermocouple_diff.mean():.4f}, std = {thermocouple_diff.std():.4f}")
```

```python
#графики acf/pacf для продифференцированного ряда
plt.figure(figsize=(12, 6))

plt.subplot(1, 2, 1)
plot_acf(thermocouple_diff, ax=plt.gca(), lags=40)
plt.title('ACF продифференцированного ряда')

plt.subplot(1, 2, 2)
plot_pacf(thermocouple_diff, ax=plt.gca(), lags=40)
plt.title('PACF продифференцированного ряда')

plt.tight_layout()
plt.show()

print("acf: быстрое затухание - признак стационарности")
print("pacf: резкий обрыв после нескольких лагов")
print("ряд стационарен")
```

```python
from statsmodels.tsa.arima.model import ARIMA
import warnings
warnings.filterwarnings('ignore')

thermocouple_clean = thermocouple.dropna().reset_index(drop=True)

train_size = int(len(thermocouple_clean) * 0.7)
train = thermocouple_clean[:train_size]
test = thermocouple_clean[train_size:]

print(f"Обучающая выборка: {len(train)} точек")
print(f"Тестовая выборка: {len(test)} точек")

best_order = (0, 1, 1)

model = ARIMA(train, order=best_order)
fitted_model = model.fit()

print(f"Параметры модели ARIMA{best_order}:")
print(fitted_model.summary())

forecast_steps = len(test)
forecast = fitted_model.forecast(steps=forecast_steps)

forecast_index = range(len(train), len(train) + len(forecast))

plt.figure(figsize=(12, 6))
plt.plot(train.index, train, label='Обучающая выборка')
plt.plot(test.index, test, label='Тестовая выборка', color='gray')
plt.plot(forecast_index, forecast, label='Прогноз ARIMA(0,1,1)', color='red')
plt.title('Прогноз ARIMA(0,1,1)')
plt.legend()
plt.grid(True)
plt.show()

mae = mean_absolute_error(test, forecast)
rmse = np.sqrt(mean_squared_error(test, forecast))

print(f"\nОЦЕНКА КАЧЕСТВА ПРОГНОЗА:")
print(f"MAE: {mae:.6f}")
print(f"RMSE: {rmse:.6f}")

naive_forecast = np.full(len(test), train.iloc[-1])
naive_mae = mean_absolute_error(test, naive_forecast)
naive_rmse = np.sqrt(mean_squared_error(test, naive_forecast))

print(f"\nСРАВНЕНИЕ С НАИВНЫМ ПРОГНОЗОМ:")
print(f"ARIMA MAE: {mae:.6f} vs Наивный MAE: {naive_mae:.6f}")
print(f"ARIMA RMSE: {rmse:.6f} vs Наивный RMSE: {naive_rmse:.6f}")

if mae < naive_mae:
    print("Модель ARIMA превосходит наивный прогноз")
else:
    print("Наивный прогноз показал лучшие результаты")

print(f"АНАЛИЗ ДЛЯ КОРОТКОГО ГОРИЗОНТА ПРОГНОЗИРОВАНИЯ (10 шагов):")
short_forecast = fitted_model.forecast(steps=10)
short_test = test[:10]

short_mae = mean_absolute_error(short_test, short_forecast)
short_rmse = np.sqrt(mean_squared_error(short_test, short_forecast))

print(f"MAE на 10 шагах: {short_mae:.6f}")
print(f"RMSE на 10 шагах: {short_rmse:.6f}")

plt.figure(figsize=(12, 6))
plt.plot(train.index[-50:], train[-50:], label='Конец обучающей выборки')
plt.plot(test.index[:20], test[:20], label='Начало тестовой выборки', color='gray')
plt.plot(range(len(train), len(train)+10), short_forecast, label='Прогноз на 10 шагов', color='red', marker='o')
plt.title('Короткий прогноз ARIMA(0,1,1) - 10 шагов')
plt.legend()
plt.grid(True)
plt.show()
```

```python
#Используйте Feature-Engineering для создания признаков (лаги, скользящие статистики, признаки даты/времени, etc)
df_features = data[['datetime', 'Thermocouple']].copy()
df_features['datetime'] = pd.to_datetime(df_features['datetime'], dayfirst=True)
df_features = df_features.sort_values('datetime').reset_index(drop=True)


df_features['Thermocouple'] = pd.to_numeric(df_features['Thermocouple'], errors='coerce')
df_features = df_features.dropna()

print(f"Исходный размер данных: {len(df_features)}")

#признаки даты и времени
df_features['hour'] = df_features['datetime'].dt.hour
df_features['minute'] = df_features['datetime'].dt.minute
df_features['second'] = df_features['datetime'].dt.second
df_features['time_of_day'] = df_features['hour'] + df_features['minute']/60 + df_features['second']/3600

#лаговые признаки
for lag in [1, 2, 3, 5, 10]:
    df_features[f'lag_{lag}'] = df_features['Thermocouple'].shift(lag)

#скользящие статистики
windows = [5, 10, 20]
for window in windows:
    df_features[f'rolling_mean_{window}'] = df_features['Thermocouple'].rolling(window=window).mean()
    df_features[f'rolling_std_{window}'] = df_features['Thermocouple'].rolling(window=window).std()
    df_features[f'rolling_min_{window}'] = df_features['Thermocouple'].rolling(window=window).min()
    df_features[f'rolling_max_{window}'] = df_features['Thermocouple'].rolling(window=window).max()

#признаки тренда и изменений
df_features['diff_1'] = df_features['Thermocouple'].diff()
df_features['diff_2'] = df_features['Thermocouple'].diff().diff()
df_features['price_change'] = df_features['Thermocouple'] - df_features['Thermocouple'].shift(1)

#удаляем строки с NaN
df_features = df_features.dropna()

print(f"Размер данных после создания признаков: {len(df_features)}")
print(f"Количество созданных признаков: {len(df_features.columns) - 2}")  # -2 потому что исключаем datetime и Thermocouple

feature_columns = [col for col in df_features.columns if col not in ['datetime', 'Thermocouple']]
print(f"\nСозданные признаки: {feature_columns}")

plt.figure(figsize=(15, 12))

#исходный ряд
plt.subplot(3, 2, 1)
plt.plot(df_features['Thermocouple'])
plt.title('Исходный ряд')
plt.grid(True)

#лаговые признаки
plt.subplot(3, 2, 2)
plt.plot(df_features['lag_1'], label='lag_1')
plt.plot(df_features['lag_5'], label='lag_5')
plt.title('Лаговые признаки')
plt.legend()
plt.grid(True)

#скользящее среднее
plt.subplot(3, 2, 3)
plt.plot(df_features['Thermocouple'], alpha=0.5, label='Исходный')
plt.plot(df_features['rolling_mean_10'], label='Скользящее среднее (10)')
plt.title('Скользящее среднее')
plt.legend()
plt.grid(True)

#скользящее стандартное отклонение
plt.subplot(3, 2, 4)
plt.plot(df_features['rolling_std_10'])
plt.title('Скользящее стандартное отклонение (10)')
plt.grid(True)

#дифференцирование
plt.subplot(3, 2, 5)
plt.plot(df_features['diff_1'])
plt.title('Первая разность')
plt.grid(True)

#время суток
plt.subplot(3, 2, 6)
plt.plot(df_features['time_of_day'])
plt.title('Время суток (нормализованное)')
plt.grid(True)

plt.tight_layout()
plt.show()

correlation_with_target = df_features[feature_columns + ['Thermocouple']].corr()['Thermocouple'].sort_values(ascending=False)
print("КОРРЕЛЯЦИЯ ПРИЗНАКОВ С ЦЕЛЕВОЙ ПЕРЕМЕННОЙ:")
for feature, corr in correlation_with_target.items():
    if feature != 'Thermocouple':
        print(f"{feature}: {corr:.4f}")

#подготовка данных для моделирования
X = df_features[feature_columns]
y = df_features['Thermocouple']

#разделение на обучающую и тестовую
train_size = int(len(X) * 0.7)
X_train, X_test = X[:train_size], X[train_size:]
y_train, y_test = y[:train_size], y[train_size:]

print(f"РАЗМЕРНОСТИ ДАННЫХ ДЛЯ МОДЕЛИРОВАНИЯ:")
print(f"X_train: {X_train.shape}, X_test: {X_test.shape}")
print(f"y_train: {y_train.shape}, y_test: {y_test.shape}")
```

```python
best_features = [
    'rolling_mean_5', 'rolling_min_5', 'rolling_max_5',
    'rolling_mean_10', 'rolling_min_10', 'rolling_max_10',
    'lag_1', 'lag_2', 'lag_3', 'lag_5', 'lag_10'
]

features_to_drop = ['hour', 'minute', 'second', 'time_of_day', 'rolling_std_5',
                   'rolling_std_10', 'rolling_std_20', 'diff_2']

X_optimized = X.drop(columns=features_to_drop)

print(f"оптимизированная матрица признаков: {X_optimized.shape}")

X_train_opt, X_test_opt = X_optimized[:train_size], X_optimized[train_size:]
```

```python
from sklearn.ensemble import RandomForestRegressor
from xgboost import XGBRegressor
from sklearn.metrics import mean_absolute_error, mean_squared_error

print("обучение рандомного леса")

rf_model = RandomForestRegressor(
    n_estimators=100,
    max_depth=10,
    random_state=42,
    n_jobs=-1
)

rf_model.fit(X_train_opt, y_train)
rf_predictions = rf_model.predict(X_test_opt)

rf_mae = mean_absolute_error(y_test, rf_predictions)
rf_rmse = np.sqrt(mean_squared_error(y_test, rf_predictions))

print(f"Random Forest MAE: {rf_mae:.6f}")
print(f"Random Forest RMSE: {rf_rmse:.6f}")

print("обучение градиентного бустинга")

xgb_model = XGBRegressor(
    n_estimators=100,
    max_depth=6,
    learning_rate=0.1,
    random_state=42,
    n_jobs=-1
)

xgb_model.fit(X_train_opt, y_train)
xgb_predictions = xgb_model.predict(X_test_opt)

xgb_mae = mean_absolute_error(y_test, xgb_predictions)
xgb_rmse = np.sqrt(mean_squared_error(y_test, xgb_predictions))

print(f"XGBoost MAE: {xgb_mae:.6f}")
print(f"XGBoost RMSE: {xgb_rmse:.6f}")

print("сравнение")
print(f"наивный прогноз MAE: {naive_mae:.6f}")
print(f"ARIMA MAE: {mae:.6f}")
print(f"Random Forest MAE: {rf_mae:.6f}")
print(f"XGBoost MAE: {xgb_mae:.6f}")

plt.figure(figsize=(15, 10))

plt.subplot(2, 2, 1)
plt.plot(y_test.values, label='Фактические значения')
plt.plot(rf_predictions, label='Random Forest', alpha=0.7)
plt.title(f'Random Forest прогноз\nMAE: {rf_mae:.6f}')
plt.legend()
plt.grid(True)

plt.subplot(2, 2, 2)
plt.plot(y_test.values, label='Фактические значения')
plt.plot(xgb_predictions, label='XGBoost', alpha=0.7)
plt.title(f'XGBoost прогноз\nMAE: {xgb_mae:.6f}')
plt.legend()
plt.grid(True)

plt.subplot(2, 1, 2)
plt.plot(y_test.values[:50], label='Фактические значения', linewidth=2)
plt.plot(rf_predictions[:50], label='Random Forest', alpha=0.8)
plt.plot(xgb_predictions[:50], label='XGBoost', alpha=0.8)
plt.title('Сравнение прогнозов (первые 50 точек)')
plt.legend()
plt.grid(True)

plt.tight_layout()
plt.show()

print("важность признаков")


rf_importance = pd.DataFrame({
    'feature': X_optimized.columns,
    'importance': rf_model.feature_importances_
}).sort_values('importance', ascending=False)

print("Random Forest - Топ-10 важных признаков:")
for i, row in rf_importance.head(10).iterrows():
    print(f"  {row['feature']}: {row['importance']:.4f}")


xgb_importance = pd.DataFrame({
    'feature': X_optimized.columns,
    'importance': xgb_model.feature_importances_
}).sort_values('importance', ascending=False)

print("\nXGBoost - Топ-10 важных признаков:")
for i, row in xgb_importance.head(10).iterrows():
    print(f"  {row['feature']}: {row['importance']:.4f}")


fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(15, 6))


rf_importance.head(10).plot.bar(x='feature', y='importance', ax=ax1)
ax1.set_title('Важность признаков - Random Forest')
ax1.tick_params(axis='x', rotation=45)


xgb_importance.head(10).plot.bar(x='feature', y='importance', ax=ax2)
ax2.set_title('Важность признаков - XGBoost')
ax2.tick_params(axis='x', rotation=45)

plt.tight_layout()
plt.show()
```
