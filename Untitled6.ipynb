# Оптимальные типы данных для каждого столбца:
# price - int32 (диапазон значений от 1.75M до 13.3M, помещается в int32)
# area - int32 (диапазон значений от 1650 до 16200, помещается в int32)
# bedrooms - int8 (диапазон 1-6)
# bathrooms - int8 (диапазон 1-4)
# stories - int8 (диапазон 1-4)
# mainroad - category (2 уникальных значения)
# guestroom - category (2 уникальных значения)
# basement - category (2 уникальных значения)
# hotwaterheating - category (2 уникальных значения)
# airconditioning - category (2 уникальных значения)
# parking - int8 (диапазон 0-3)
# prefarea - category (2 уникальных значения)
# furnishingstatus - category (3 уникальных значения)

```python
df_housing_optimized = df_housing.copy()
df_housing_optimized['price'] = df_housing_optimized['price'].astype('int32')
df_housing_optimized['area'] = df_housing_optimized['area'].astype('int32')
df_housing_optimized['bedrooms'] = df_housing_optimized['bedrooms'].astype('int8')
df_housing_optimized['bathrooms'] = df_housing_optimized['bathrooms'].astype('int8')
df_housing_optimized['stories'] = df_housing_optimized['stories'].astype('int8')
df_housing_optimized['parking'] = df_housing_optimized['parking'].astype('int8')

categorical_columns = ['mainroad', 'guestroom', 'basement', 'hotwaterheating', 'airconditioning', 'prefarea', 'furnishingstatus']
for col in categorical_columns:
    df_housing_optimized[col] = df_housing_optimized[col].astype('category')
```

```python
(df_housing_optimized.dtypes)
```

```python
print(f"\nиспользование памяти после оптимизации: {df_housing_optimized.memory_usage(deep=True).sum() / 1024:.2f} KB")
```
