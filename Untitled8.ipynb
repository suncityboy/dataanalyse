Поиск аномалий

Статистические эвристики

```python

def detect_anomalies_iqr(data, columns, multiplier=1.5):

    #обнаружение аномалий методом iqr
    #multiplier: 1.5 для обычных выбросов, 3 для экстремальных

    anomalies_mask = pd.Series([False] * len(data))

    for column in columns:
        q1 = data[column].quantile(0.25)
        q3 = data[column].quantile(0.75)
        iqr = q3 - q1

        lower_bound = q1 - multiplier * iqr
        upper_bound = q3 + multiplier * iqr


        column_anomalies = (data[column] < lower_bound) | (data[column] > upper_bound)
        anomalies_mask = anomalies_mask | column_anomalies

        print(f"{column}: границы [{lower_bound:.4f}, {upper_bound:.4f}]")

    return anomalies_mask

numeric_columns = ['Current', 'Pressure', 'Temperature', 'Voltage',
                   'Accelerometer1RMS', 'Accelerometer2RMS', 'Volume Flow RateRMS']

print("обнаружение аномалий методом iqr (multiplier=1.5):")
anomalies_1_5 = detect_anomalies_iqr(df, numeric_columns, multiplier=1.5)
print('_')
print("обнаружение аномалий методом iqr (multiplier=3):")
anomalies_3 = detect_anomalies_iqr(df, numeric_columns, multiplier=3)
print('_')
df['anomaly_iqr_1.5'] = anomalies_1_5.astype(int)
df['anomaly_iqr_3'] = anomalies_3.astype(int)

print(f"аномалий по iqr 1.5: {anomalies_1_5.sum()} ({anomalies_1_5.sum()/len(df)*100:.2f}%)")
print(f"аномалий по iqr 3: {anomalies_3.sum()} ({anomalies_3.sum()/len(df)*100:.2f}%)")
print(f"исходных аномалий: {df['anomaly'].sum()} ({df['anomaly'].sum()/len(df)*100:.2f}%)")
```

Кластеризация

```python
df_fe = df.copy()
df_fe['hour'] = df_fe['datetime'].dt.hour
df_fe['minute'] = df_fe['datetime'].dt.minute
df_fe['second'] = df_fe['datetime'].dt.second
df_fe['time_of_day'] = df_fe['hour'] * 3600 + df_fe['minute'] * 60 + df_fe['second']

print("созданные признаки даты/времени:")
print(df_fe[['datetime', 'hour', 'minute', 'second', 'time_of_day']].head())
```

```python
#лаги
lag_columns = ['Current', 'Pressure', 'Temperature', 'Voltage']
lags = [1, 3, 5]

for column in lag_columns:
    for lag in lags:
        lag_values = df_fe[column].shift(lag)
        mean_val = df_fe[column].mean()
        df_fe[f'{column}_lag_{lag}'] = lag_values.fillna(mean_val)

print("лаги созданы для столбцов:", lag_columns)
print(f"добавлено {len(lag_columns) * len(lags)} лагов")
```

```python
#скользящие статистики
window_sizes = [5, 10, 15]

for column in lag_columns:
    for window in window_sizes:
        df_fe[f'{column}_ma_{window}'] = df_fe[column].rolling(window=window, min_periods=1).mean()
        df_fe[f'{column}_std_{window}'] = df_fe[column].rolling(window=window, min_periods=1).std()
        df_fe[f'{column}_min_{window}'] = df_fe[column].rolling(window=window, min_periods=1).min()
        df_fe[f'{column}_max_{window}'] = df_fe[column].rolling(window=window, min_periods=1).max()

print(f"добавлено {len(lag_columns) * len(window_sizes) * 4} скользящих признаков")
```

```python
#разности
for column in lag_columns:
    diff_1 = df_fe[column].diff(1)
    diff_3 = df_fe[column].diff(3)
    df_fe[f'{column}_diff_1'] = diff_1.fillna(0)
    df_fe[f'{column}_diff_3'] = diff_3.fillna(0)
```

```python
#взаимодействия признаков
df_fe['current_pressure_ratio'] = df_fe['Current'] / (df_fe['Pressure'] + 1e-6)
df_fe['temperature_voltage_product'] = df_fe['Temperature'] * df_fe['Voltage']
df_fe['power_estimate'] = df_fe['Current'] * df_fe['Voltage']
```

```python
#ЕМА
for column in lag_columns:
    df_fe[f'{column}_ema_5'] = df_fe[column].ewm(span=5, adjust=False).mean()
    df_fe[f'{column}_ema_10'] = df_fe[column].ewm(span=10, adjust=False).mean()
```

```python
#k-means кластеризация
from sklearn.cluster import KMeans
from sklearn.preprocessing import StandardScaler
import numpy as np

features_for_clustering = ['Current', 'Pressure', 'Temperature', 'Voltage',
                          'Accelerometer1RMS', 'Accelerometer2RMS', 'Volume Flow RateRMS']

X = df_fe[features_for_clustering].copy()
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

print("данные подготовлены для кластеризации")
print(f"размерность данных: {X_scaled.shape}")
```

```python
kmeans = KMeans(n_clusters=3, random_state=42, n_init=10)
kmeans_labels = kmeans.fit_predict(X_scaled)

distances = kmeans.transform(X_scaled)
min_distances = np.min(distances, axis=1)

threshold_kmeans = np.percentile(min_distances, 95)
df_fe['anomaly_kmeans'] = (min_distances > threshold_kmeans).astype(int)

print(f"обнаружено аномалий: {df_fe['anomaly_kmeans'].sum()}")
print(f"процент аномалий: {df_fe['anomaly_kmeans'].sum()/len(df_fe)*100:.2f}%")
print(f"пороговое расстояние: {threshold_kmeans:.3f}")
```

```python
plt.figure(figsize=(15, 10))

anomalies_kmeans = df_fe[df_fe['anomaly_kmeans'] == 1]

plt.subplot(2, 2, 1)
plt.plot(df_fe['datetime'], df_fe['Current'], 'b-', alpha=0.7, linewidth=0.8, label='норма')
if not anomalies_kmeans.empty:
    plt.scatter(anomalies_kmeans['datetime'], anomalies_kmeans['Current'],
                color='red', s=30, label='k-means аномалии')
plt.title('ток с k-means аномалиями')
plt.ylabel('а')
plt.grid(True, alpha=0.3)
plt.legend()

plt.subplot(2, 2, 2)
plt.plot(df_fe['datetime'], df_fe['Pressure'], 'b-', alpha=0.7, linewidth=0.8)
if not anomalies_kmeans.empty:
    plt.scatter(anomalies_kmeans['datetime'], anomalies_kmeans['Pressure'],
                color='red', s=30)
plt.title('давление с k-means аномалиями')
plt.ylabel('бар')
plt.grid(True, alpha=0.3)

plt.subplot(2, 2, 3)
plt.plot(df_fe['datetime'], df_fe['Temperature'], 'b-', alpha=0.7, linewidth=0.8)
if not anomalies_kmeans.empty:
    plt.scatter(anomalies_kmeans['datetime'], anomalies_kmeans['Temperature'],
                color='red', s=30)
plt.title('температура с k-means аномалиями')
plt.ylabel('°C')
plt.grid(True, alpha=0.3)

plt.subplot(2, 2, 4)
plt.plot(df_fe['datetime'], df_fe['Voltage'], 'b-', alpha=0.7, linewidth=0.8)
if not anomalies_kmeans.empty:
    plt.scatter(anomalies_kmeans['datetime'], anomalies_kmeans['Voltage'],
                color='red', s=30)
plt.title('напряжение с k-means аномалиями')
plt.ylabel('в')
plt.grid(True, alpha=0.3)

plt.tight_layout()
plt.show()
```

```python
#оценка качества
if 'anomaly' in df_fe.columns:
    from sklearn.metrics import confusion_matrix, precision_score, recall_score, f1_score

    print("оценка качества k-means кластеризации:")

    y_true = df_fe['anomaly']
    y_pred_kmeans = df_fe['anomaly_kmeans']

    cm_kmeans = confusion_matrix(y_true, y_pred_kmeans)

    plt.figure(figsize=(6, 5))
    sns.heatmap(cm_kmeans, annot=True, fmt='d', cmap='Blues',
                xticklabels=['норма', 'аномалия'],
                yticklabels=['норма', 'аномалия'])
    plt.title('confusion matrix: k-means')
    plt.ylabel('истинный класс')
    plt.xlabel('предсказанный класс')
    plt.show()

    precision = precision_score(y_true, y_pred_kmeans)
    recall = recall_score(y_true, y_pred_kmeans)
    f1 = f1_score(y_true, y_pred_kmeans)

    print(f"precision: {precision:.3f}")
    print(f"recall: {recall:.3f}")
    print(f"f1-score: {f1:.3f}")

    if 'anomaly_iqr_1.5' in df_fe.columns and 'anomaly_iqr_3' in df_fe.columns:
        print("\nсравнение с iqr методами:")
        y_pred_iqr_1_5 = df_fe['anomaly_iqr_1.5']
        y_pred_iqr_3 = df_fe['anomaly_iqr_3']

        print(f"iqr 1.5: precision={precision_score(y_true, y_pred_iqr_1_5):.3f}, recall={recall_score(y_true, y_pred_iqr_1_5):.3f}, f1={f1_score(y_true, y_pred_iqr_1_5):.3f}")
        print(f"iqr 3: precision={precision_score(y_true, y_pred_iqr_3):.3f}, recall={recall_score(y_true, y_pred_iqr_3):.3f}, f1={f1_score(y_true, y_pred_iqr_3):.3f}")
        print(f"k-means: precision={precision:.3f}, recall={recall:.3f}, f1={f1:.3f}")
```

Isolation Forest

```python
#isolation forest
from sklearn.ensemble import IsolationForest
from sklearn.preprocessing import StandardScaler

features_for_isolation = ['Current', 'Pressure', 'Temperature', 'Voltage',
                         'Accelerometer1RMS', 'Accelerometer2RMS', 'Volume Flow RateRMS']

X_iso = df_fe[features_for_isolation].copy()
scaler_iso = StandardScaler()
X_iso_scaled = scaler_iso.fit_transform(X_iso)
```

```python
iso_forest = IsolationForest(contamination=0.05, random_state=42, n_estimators=100)
iso_predictions = iso_forest.fit_predict(X_iso_scaled)

df_fe['anomaly_isolation'] = (iso_predictions == -1).astype(int)
df_fe['isolation_score'] = iso_forest.decision_function(X_iso_scaled)

print(f"обнаружено аномалий: {df_fe['anomaly_isolation'].sum()}")
print(f"процент аномалий: {df_fe['anomaly_isolation'].sum()/len(df_fe)*100:.2f}%")
```

```python
methods_comparison = []

if 'anomaly' in df_fe.columns:
    y_true = df_fe['anomaly']

    anomaly_methods = {
        'iqr_1.5': 'anomaly_iqr_1.5',
        'iqr_3': 'anomaly_iqr_3',
        'kmeans': 'anomaly_kmeans',
        'isolation_forest': 'anomaly_isolation'
    }

    for method_name, col_name in anomaly_methods.items():
        if col_name in df_fe.columns:
            y_pred = df_fe[col_name]

            precision = precision_score(y_true, y_pred)
            recall = recall_score(y_true, y_pred)
            f1 = f1_score(y_true, y_pred)
            accuracy = (y_pred == y_true).mean()

            n_anomalies = y_pred.sum()
            pct_anomalies = n_anomalies / len(y_pred) * 100

            tn, fp, fn, tp = confusion_matrix(y_true, y_pred).ravel()

            methods_comparison.append({
                'method': method_name,
                'precision': precision,
                'recall': recall,
                'f1_score': f1,
                'accuracy': accuracy,
                'true_positives': tp,
                'false_positives': fp,
                'true_negatives': tn,
                'false_negatives': fn,
                'n_anomalies': n_anomalies,
                'pct_anomalies': pct_anomalies
            })

    results_df = pd.DataFrame(methods_comparison)

    results_df = results_df.sort_values('f1_score', ascending=False)

    print("агрегированные результаты всех методов:")
    print(results_df.round(4))

else:
    anomaly_counts = {}
    for col in df_fe.columns:
        if 'anomaly' in col and col != 'anomaly':
            anomaly_counts[col] = df_fe[col].sum()

    results_df = pd.DataFrame(list(anomaly_counts.items()), columns=['method', 'n_anomalies'])
    results_df['pct_anomalies'] = results_df['n_anomalies'] / len(df_fe) * 100

    print("сравнение количества обнаруженных аномалий:")
    print(results_df.round(4))
```

```python
if 'anomaly' in df_fe.columns and len(methods_comparison) > 0:
    metrics_to_plot = ['precision', 'recall', 'f1_score', 'accuracy']

    plt.figure(figsize=(15, 10))

    for i, metric in enumerate(metrics_to_plot, 1):
        plt.subplot(2, 2, i)
        sorted_df = results_df.sort_values(metric, ascending=True)
        plt.barh(sorted_df['method'], sorted_df[metric], color='skyblue', alpha=0.7)
        plt.xlabel(metric)
        plt.title(f'сравнение методов по {metric}')

        for j, v in enumerate(sorted_df[metric]):
            plt.text(v + 0.01, j, f'{v:.3f}', va='center')

        plt.grid(True, alpha=0.3, axis='x')

    plt.tight_layout()
    plt.show()
```
